```python
# Block 2: Import libraries and set the base path
import os
import glob
import pandas as pd

# Pointing to the specific 'home_and_away' folder structure you shared
base_path = "/content/datasets/league_table/home_and_away"
all_gameweek_data = []

print("✅ Setup complete!")
```

```python
# Block 3: Loop through years and read every gameweek CSV
for year in range(2016, 2025):
    folder_name = f"gameweek_{year}"
    folder_path = os.path.join(base_path, folder_name)

    # Check if the folder exists
    if os.path.exists(folder_path):
        # Get all CSV files in this specific folder
        csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

        for file in csv_files:
            try:
                # Read the CSV
                df = pd.read_csv(file)

                # Create a Season column (e.g., 2016-2017)
                df['Season'] = f"{year}-{year+1}"

                # Extract the gameweek number from the filename (e.g., '1.csv' -> '1')
                filename = os.path.basename(file)
                df['Gameweek'] = filename.split('.')[0]

                # Append to our list
                all_gameweek_data.append(df)
            except Exception as e:
                print(f"Error reading {file}: {e}")

print(f"✅ Successfully read {len(all_gameweek_data)} CSV files.")
```

```python
# Block 4: Combine everything and clean the data

# Merge all dataframes into one large dataframe
master_df = pd.concat(all_gameweek_data, ignore_index=True)

# Drop completely empty rows (if any)
master_df.dropna(how='all', inplace=True)

# Standardize column names (strip spaces, replace spaces with underscores)
master_df.columns = master_df.columns.str.strip().str.replace(' ', '_')

# Show the total row count and preview the first 5 rows
print(f"✅ Total rows in merged dataset: {len(master_df)}")
display(master_df.head())
```

```python
# Block 5: Save as CSV and download to your Mac
from google.colab import files

file_name = "master_premier_league.csv"

# Save to Colab's temporary storage
master_df.to_csv(f"/content/{file_name}", index=False)
print("✅ Saved locally in Colab. Starting download to your Mac...")

# Trigger download to your actual computer
files.download(f"/content/{file_name}")
```

```python
# Block 6: Validate the Badge URLs
import requests

# 1. Get a list of unique URLs (we don't need to test Arsenal 380 times)
unique_urls = master_df['badge_url'].dropna().unique()
print(f"Found {len(unique_urls)} unique team badges. Testing a sample of 5...\n")

# 2. Loop through the first 5 URLs and check their web status
for url in unique_urls[:5]:
    try:
        # Ping the URL to see if it responds
        response = requests.get(url, timeout=5)

        if response.status_code == 200:
            print(f"✅ VALID: The image loaded successfully!")
            print(f"   URL: {url}\n")
        else:
            print(f"❌ BROKEN: Returns status code {response.status_code}")
            print(f"   URL: {url}\n")

    except Exception as e:
        print(f"⚠️ ERROR: Could not connect to {url}")
```
